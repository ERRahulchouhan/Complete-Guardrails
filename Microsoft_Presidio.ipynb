# Microsoft Presidio

Detect and anonymize common PII such as email addresses and phone numbers. Use synthetic data in examples. For production, review the recognizers and language models for your region and data types.

Install when ready: `%pip install presidio-analyzer presidio-anonymizer`. Presidio may also require a compatible spaCy language model.

```python
from presidio_analyzer import AnalyzerEngine
from presidio_anonymizer import AnonymizerEngine

text = "Contact Alex at alex@example.com or 555-010-1234 about the test request."
analyzer = AnalyzerEngine()
anonymizer = AnonymizerEngine()

findings = analyzer.analyze(
    text=text,
    entities=["EMAIL_ADDRESS", "PHONE_NUMBER"],
    language="en",
)
print("Detected entities:", [(item.entity_type, item.start, item.end) for item in findings])
result = anonymizer.anonymize(text=text, analyzer_results=findings)
print("Anonymized text:", result.text)
```
