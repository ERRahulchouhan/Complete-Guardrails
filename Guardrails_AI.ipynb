# Guardrails AI

Validate an output with a length validator. Install the package and validator from the Guardrails Hub only when you are ready to run the notebook.

Setup: `%pip install guardrails-ai`, then in a terminal run `guardrails hub install hub://guardrails/valid_length`. This validation example does not call an LLM.

```python
from guardrails import Guard, OnFailAction
from guardrails.hub import ValidLength

output_guard = Guard().use(
    ValidLength,
    min=1,
    max=120,
    on_fail=OnFailAction.EXCEPTION,
)

candidate_output = "Your refund request has been received."
result = output_guard.validate(candidate_output)
print("Validated output:", result.validated_output)
print("Validation passed:", result.validation_passed)
```
