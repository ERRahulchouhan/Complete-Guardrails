# garak

Run an automated probe against a model endpoint you own or are authorized to test. Start with a small probe set and review the current garak CLI options for your installed release.

Install when ready: `%pip install garak`. Set any provider credentials in the environment. This launches network requests to the configured provider.

```python
import shutil
import subprocess

if shutil.which("garak") is None:
    raise RuntimeError("Install garak in this environment before running this cell.")

command = [
    "garak",
    "--model_type", "openai",
    "--model_name", "gpt-4o-mini",
    "--probes", "promptinject",
]
completed = subprocess.run(command, check=False, text=True)
print("garak exit code:", completed.returncode)
```
